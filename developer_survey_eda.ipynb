## Survey Response Completeness

Identify respondents with non-missing answers to all survey-question columns listed in the schema and present in the dataset.


```python
schema = pd.read_csv('/content/drive/MyDrive/stack-overflow-developer-survey-2025/survey_results_schema.csv')

questions = set(schema['qname'])

available_questions = questions.intersection(Stack_Overflow.columns)

df_questions = Stack_Overflow[list(available_questions)]

complete_mask = df_questions.notna().all(axis=1)

complete_responses = complete_mask.sum()

print(complete_responses)
```

## Professional Experience: Descriptive Statistics

Calculate the mean, median, and mode of respondents' professional work experience (`WorkExp`).


```python
workexp_mean = Stack_Overflow['WorkExp'].mean()
workexp_median = Stack_Overflow['WorkExp'].median()
workexp_mode = Stack_Overflow['WorkExp'].mode()

result = pd.Series({
    'mean': workexp_mean,
    'median': workexp_median,
    'mode': workexp_mode.iloc[0] if not workexp_mode.empty else None
})

print(result)
```

## Python Adoption Among Respondents

Calculate the percentage of respondents who reported using Python among their programming languages.


```python
python_mask = Stack_Overflow['LanguageHaveWorkedWith'].str.contains('Python', na=False)

python_percent = python_mask.mean() * 100

print(f"{python_percent:.1f}%")
```

## Learning to Code Through Online Courses

Count respondents who selected online courses as a way of learning to code.

## Python Developer Compensation by Country

Compare the mean and median annual compensation of Python developers across countries.


```python
python_mask = Stack_Overflow['LanguageHaveWorkedWith'].str.contains('Python', na=False)

df_python = Stack_Overflow[python_mask][['Country', 'ConvertedCompYearly']]

df_python = df_python.dropna(subset=['ConvertedCompYearly'])

result = df_python.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median'])

print(result)
```

## Education of the Five Highest-Paid Respondents

Identify the education levels of the five respondents with the highest reported annual compensation.


```python
top5_education = (
    Stack_Overflow
    .dropna(subset=['ConvertedCompYearly'])
    .sort_values(by='ConvertedCompYearly', ascending=False)
    .head(5)[['EdLevel']]
)

print(top5_education)
```

## Python Adoption Across Age Groups

Calculate the percentage of Python users within each age category.


```python
python_mask = Stack_Overflow['LanguageHaveWorkedWith'].str.contains('Python', na=False)

age_python_percent = (
    Stack_Overflow
    .assign(PythonUser=python_mask)
    .groupby('Age')['PythonUser']
    .mean()
    .mul(100)
    .round(2)
    .astype(str) + '%'
)

print(age_python_percent)
```

## Industries of High-Paid Remote Professionals

Identify the most common industries among remote respondents with annual compensation at or above the 75th percentile.


```python
salary_75 = Stack_Overflow['ConvertedCompYearly'].quantile(0.75)

high_paid_remote = Stack_Overflow[
    (Stack_Overflow['ConvertedCompYearly'] >= salary_75) &
    (Stack_Overflow['RemoteWork'].str.contains('Remote', na=False))
]

top_industries = high_paid_remote['Industry'].dropna().value_counts()

print(top_industries)
```
